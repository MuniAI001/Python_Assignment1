```
## Create a class with function that checks whether the given number is Odd or Even
class OddEven():
  def OddEven():
    inp=int(input('Enter a number: '))
    if inp%2==0:
      print(f'{inp} is even number')
    else:
      print(f'{inp} is odd number')

OddEven.OddEven()
```

```
'''Create a function that tells elegibility of marriage for male and female according to thei
r age limit like 21 for male and 18 for female'''

class ElegiblityForMarriage():
  def Elegible(genter,age):
    if genter.lower() == "male" and age > 24:
      print("Eligible")
    elif genter.lower() == "female" and age > 21:
      print("Eligible")
    else:
      print("Not Eligible")

inp1=input("Your genter: ")
inp2=int(input("Your Age: "))

ElegiblityForMarriage.Elegible(genter=inp1,age=inp2)
```

```
#calculate the percentage of your 10th mark

class FindPercent():
  def percentage(*Args):
    Total = 0
    for i in Args:
      Total = Total + i
    print("Total :",Total)
    print("Percentage : ", Total/5)

Subject1=int(input("Subject1 :"))
Subject2=int(input("Subject2 :"))
Subject3=int(input("Subject3 :"))
Subject4=int(input("Subject4 :"))
Subject5=int(input("Subject5 :"))

Marks=[Subject1,Subject2,Subject3,Subject4,Subject5]
FindPercent.percentage(*Marks)
```
