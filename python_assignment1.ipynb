##Assignment3: #Body Mass Index

```
'''
Expected output:

Enter the BMI Index:34
Very Overweight
'''
bmi = int(input("Enter the BMI Index:"))
if bmi < 18.5:
  print("Underweight")
elif bmi <= 24.9:
  print("Normal")
elif bmi <= 29.9:
  print ("Overweight")
else:
  print("Very overweight")
```
